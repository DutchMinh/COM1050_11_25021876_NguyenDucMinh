```python
#w8a2
class Date:
    def __init__(self, day, month, year):
        self.day = day
        self.month = month
        self.year = year

    def is_valid(self):
        if self.year < 1:
            return False
        if self.month < 1 or self.month > 12:
            return False
        if self.month in [1, 3, 5, 7, 8, 10, 12]:
            max_day = 31
        elif self.month in [4, 6, 9, 11]:
            max_day = 30
        else:  
            leap = (self.year % 400 == 0 or (self.year % 4 == 0 and self.year % 100 != 0))
            max_day = 29 if leap else 28
        return 1 <= self.day <= max_day

    def next_day(self):
        if self.month in [1, 3, 5, 7, 8, 10, 12]:
            max_day = 31
        elif self.month in [4, 6, 9, 11]:
            max_day = 30
        else:
            leap = (self.year % 400 == 0 or (self.year % 4 == 0 and self.year % 100 != 0))
            max_day = 29 if leap else 28
        self.day += 1

        if self.day > max_day:
            self.day = 1
            self.month += 1

            if self.month > 12:
                self.month = 1
                self.year += 1
s = input().strip()
d,m,y = map(int, s.split('/'))
date = Date(d,m,y)
if not date.is_valid():
    print('INVALID')
else:
    date.next_day()
    print(f"{date.day:02d}/{date.month:02d}/{date.year}")
```

```python
#w8a3
class Point:
    def __init__(self, x_value, y_value):
        self.x = x_value
        self.y = y_value
    
    def is_origin(self):
        return self.x == 0 and self.y == 0
    
    def on_which_axis(self):
        if self.x == 0:
            return "Nằm trên trục tung"
        if self.y == 0:
            return "Nằm trên trục hoành"
        else:
            return "Không nằm trên trục tọa độ"
    def dist(self):
        return ((self.x)**2 + (self.y)**2)**0.5
```

```python
#W8a4
class Calculator:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def add(self):
        print(f"{self.x} + {self.y} = {self.x + self.y}")
    def subtract(self):
        print(f"{self.x} - {self.y} = {self.x - self.y}")
    def product(self):
        print(f"{self.x} * {self.y} = {self.x * self.y}")
    def divide(self):
        print(f"{self.x} / {self.y} = {self.x / self.y}")
    def power(self):
        print(f"{self.x} ^ {self.y} = {self.x ** self.y}")
    def mod(self):
        print(f"{self.x} % {self.y} = {self.x % self.y}")
    def set_numbers(self, new_x, new_y):
        self.x = new_x
        self.y = new_y
Continue = True
x, y = map(float, input("Nhập hai số vào để tính theo thứ tự:").split())
num = Calculator(x, y)
while Continue:
    operator = input("Bạn muốn thực hiện tính toán như thế nào? (+, -, *, /, ^, %)?")
    if operator == "+":
        print(num.add())
    elif operator == "-":
        print(num.subtract())
    elif operator == "*":
        print(num.multiply())
    elif operator == "/":
        print(num.divide())
    elif operator == "^":
        print(num.power())
    elif operator == "%":
        print(num.modulo())
    else:
        print("Bạn đã nhập sai toán tử!")
    choice = input("Bạn muốn dừng chưa? (Y/N)")
    if choice == "Y":
        Continue = False
        print("Hẹn gặp lại lần sau!")
    else:
        x, y = map(int, input("Hãy nhập lại số nhé").split())
```

```python
#w8a5
class ShoppingCart:
    def __init__(self):
        self.cart = []

    def add_product(self):
        sp = input('Nhập sản phẩm: ')
        price = float(input('Nhập giá tiền của sản phẩm: '))
        self.cart.append({sp:price})
    def remove_product(self):
        w_sp = input('Nhập tên sản phẩm: ')
        xoa = False
        for i in self.cart:
            if list(i.keys())[0] == w_sp:
                self.cart.remove(i)
                xoa = True
                break
        if xoa:
            print(f'Đã xóa sản phẩm {w_sp}')
        else:
            print(f'Không tìm thấy sản phẩm {w_sp}')

    def check_cart(self):
        if self.cart != []:
            print('Giỏ hàng không bị rỗng')
        else:
            print('Giỏ hàng rỗng')
    def count_money(self):
        count = 0
        for i in self.cart:
            count += list(i.values())[0]
        print(f'Tổng số tiền của giỏ hàng: {count}')
    def show_cart(self):
        if self.cart:
            for i in self.cart:
                print(i,end = '\n')
        else:
            print(self.cart)
    def delete(self):
        self.cart = []
def menu():
    print("\n--- MENU ---") 
    print("1. Thêm một sản phẩm mới") 
    print("2. Xóa 1 sản phẩm khỏi giỏ hàng") 
    print("3. Kiểm tra giỏ hàng") 
    print("4. Tính tiền") 
    print("5. Hiển thị giỏ hàng") 
    print("6. Xóa toàn bộ") 
    
stillShopping = True
gio_hang = ShoppingCart()
while stillShopping:
    menu()
    opt = input("Nhập số trước mỗi lựa chọn: ")
    match opt:
        case '1': gio_hang.add_product()
        case '2': gio_hang.remove_product()
        case '3': gio_hang.check_cart()
        case '4': gio_hang.count_money()
        case '5': gio_hang.show_cart()
        case '6': gio_hang.delete()
    cont = input("Bạn có muốn thoát? 'y'/'n': ")
    if cont == 'y':
        stillShopping = False
```

```python
#w8a6
class Rectangle:
    def __init__(self, w, h):
        self.width = w
        self.height = h
    def area(self):
        return self.width * self.height
    def perimeter(self):
        return 2*(self.width + self.height)
    def scale(self, k):
        self.width = self.width * k
        self.height = self.height * k

w, h, k = map(float, input().split())
rec = Rectangle(w, h)
rec.scale(k)
print(rec.area(), rec.perimeter())
```

```python
#w8a7
from math import*
class Fraction:
  def __init__(self,tu_1,mau_1,oper,tu_2,mau_2):
    self.tu_1=tu_1
    self.tu_2=tu_2
    if mau_1 == 0 or mau_2 == 0:
      raise ValueError('Mẫu khác 0')
    self.mau_2=mau_2
    self.mau_1=mau_1
    self.oper = oper
  def add(self):
    if self.mau_1 == self.mau_2:
      return f'{(self.tu_1+self.tu_2)//gcd(self.tu_1+self.tu_2,self.mau_1)}/{self.mau_1}'
    else:
      bcnn = lcm(self.mau_1,self.mau_2)
      self.tu_1 = bcnn // self.mau_1
      self.tu_2 = bcnn // self.mau_2
      return f'{(self.tu_1 + self.tu_2)//gcd(self.tu_1 + self.tu_2,bcnn)}/{bcnn // gcd(self.tu_1 + self.tu_2,bcnn)}'
  def sub(self):
    if self.mau_1 == self.mau_2:
      return f'{(self.tu_1-self.tu_2)//gcd(self.tu_1-self.tu_2,self.mau_1)}/{self.mau_1}'
    else:
      bcnn = lcm(self.mau_1,self.mau_2)
      self.tu_1 = bcnn // self.mau_1
      self.tu_2 = bcnn // self.mau_2
      return f'{(self.tu_1 - self.tu_2)//gcd(self.tu_1 - self.tu_2,bcnn)}/{bcnn // gcd(self.tu_1 - self.tu_2,bcnn)}'
  def mul(self):
    return f'{(self.tu_1 * self.tu_2)//gcd(self.tu_1 * self.tu_2,self.mau_1 * self.mau_2)}/{(self.mau_1 * self.mau_2)//gcd(self.tu_1 * self.tu_2,self.mau_1 * self.mau_2)}'
  def div(self):
    return f'{(self.tu_1 * self.mau_2)//gcd(self.tu_1 * self.mau_2,self.tu_2 * self.mau_1)}/{(self.mau_1 * self.tu_2)//gcd(self.tu_1 * self.mau_2,self.tu_2 * self.mau_1)}'
  def menu(self):
    if self.oper == '+':
      print(self.add())
    elif self.oper == '-':
      print(self.sub())
    elif self.oper == '*':
      print(self.mul())
    elif self.oper == '/':
      print(self.div())

tu_1,mau_1,oper,tu_2,mau_2 = map(str,input().split())
s=Fraction(int(tu_1),int(mau_1),oper,int(tu_2),int(mau_2))
s.menu()
```

```python
#w8a8
class BankAccount:
  def __init__(self,owner,balance):
    self.owner=owner
    self.balance=balance
  def deposit(self,dep):
    self.balance += dep
  def withdraw(self,wdr):
    if self.balance >= wdr:
      self.balance -= wdr
  def __str__(self):
    return ('%.2f' %self.balance)
owner,balance = map(str,input().split())
s=BankAccount(owner,float(balance))
n=int(input())
for _ in range(n):
  method, money = map(str,input().split())
  money = float(money)
  if method == 'WITHDRAW':
    s.withdraw(money)
  elif method == 'DEPOSIT':
    s.deposit(money)
print(s)
```

```python
#w8a9
class Student:
  def __init__(self,name):
    self.name = name
    self.scores = []
  def add(self,sub,score):
    self.scores.append((sub,score))
  def avg(self):
    b=[x[1] for x in self.scores]
    return round(sum(b)/len(b),2)
  def rank(self):
    if self.avg() >=8:
      return 'Excellent'
    elif 6.5<=self.avg() < 8:
      return 'Good'
    elif 5 <= self.avg() < 6.5:
      return 'Average'
    else:
      return 'Poor'
name=input()
s=Student(name)
n=int(input())
for _ in range(n):
  sub,score = map(str,input().split())
  s.add(sub,float(score))
print(f'{name} {s.avg():.2f} {s.rank()}')
```

```python
#w8a11
from math import*
class space:
  def __init__(self,a1,a2,a3,b1,b2,b3,c1,c2,c3,d1,d2,d3):
    self.a1=a1
    self.a2=a2
    self.a3=a3
    self.b1=b1
    self.b2=b2
    self.b3=b3
    self.c1=c1
    self.c2=c2
    self.c3=c3
    self.d1=d1
    self.d2=d2
    self.d3=d3
  def mp1(self):
    n1,n2,n3=self.a1-self.b1,self.a2-self.b2,self.a3-self.b3
    k1,k2,k3=self.a1-self.c1,self.a2-self.c2,self.a3-self.c3
    return [n2*k3-k2*n3,n3*k1-k3*n1,n1*k2-k1*n2]
  def mp2(self):
    n1,n2,n3=self.c1-self.b1,self.c2-self.b2,self.c3-self.b3
    k1,k2,k3=self.d1-self.c1,self.d2-self.c2,self.d3-self.c3
    return [n2*k3-k2*n3,n3*k1-k3*n1,n1*k2-k1*n2]
  def goc(self):
    a,b,c=self.mp1()
    x,y,z=self.mp2()
    tu = abs(a*x + b*y + c*z)
    mau= sqrt(a**2 + b**2 +c**2) * sqrt(x**2 +y**2 +z**2)
    return round(degrees(acos(tu/mau)),2)
a1,a2,a3 = map(float,input().split())
b1,b2,b3 = map(float,input().split())
c1,c2,c3 = map(float,input().split())
d1,d2,d3 = map(float,input().split())
s = space(a1,a2,a3,b1,b2,b3,c1,c2,c3,d1,d2,d3)
```
