```python
#w8a2
class Date:
    def __init__(self, day, month, year):
        self.day = day
        self.month = month
        self.year = year

    def is_valid(self):
        if self.year < 1:
            return False
        if self.month < 1 or self.month > 12:
            return False
        if self.month in [1, 3, 5, 7, 8, 10, 12]:
            max_day = 31
        elif self.month in [4, 6, 9, 11]:
            max_day = 30
        else:  
            leap = (self.year % 400 == 0 or (self.year % 4 == 0 and self.year % 100 != 0))
            max_day = 29 if leap else 28
        return 1 <= self.day <= max_day

    def next_day(self):
        if self.month in [1, 3, 5, 7, 8, 10, 12]:
            max_day = 31
        elif self.month in [4, 6, 9, 11]:
            max_day = 30
        else:
            leap = (self.year % 400 == 0 or (self.year % 4 == 0 and self.year % 100 != 0))
            max_day = 29 if leap else 28
        self.day += 1

        if self.day > max_day:
            self.day = 1
            self.month += 1

            if self.month > 12:
                self.month = 1
                self.year += 1
s = input().strip()
d,m,y = map(int, input().split('/'))
date = Date(d,m,y)
if not date.is_valid():
    print('INVALID')
else:
    print(f"{date.day:02d}/{date.month:02d}/{date.year}")
```
