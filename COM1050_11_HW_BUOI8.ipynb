```python
#w8a1
pi = 3.14
class cylinder:
    def __init__(self, r, h):
        self.area = 4 * pi * r + 2 * pi * r * h
        self.volume = pi * (r ** 2) * h
r, h = map(float, input().split())
cylinder_calc = cylinder(r, h)
print(f"Area: {round(cylinder_calc.area, 2)}", f"Volume: {round(cylinder_calc.volume, 2)}", end='\n')
```

```python
#w8a2
class Date:
    def __init__(self, day, month, year):
        self.day = day
        self.month = month
        self.year = year

    def is_valid(self):
        if self.year < 1:
            return False
        if self.month < 1 or self.month > 12:
            return False
        if self.month in [1, 3, 5, 7, 8, 10, 12]:
            max_day = 31
        elif self.month in [4, 6, 9, 11]:
            max_day = 30
        else:  
            leap = (self.year % 400 == 0 or (self.year % 4 == 0 and self.year % 100 != 0))
            max_day = 29 if leap else 28
        return 1 <= self.day <= max_day

    def next_day(self):
        if self.month in [1, 3, 5, 7, 8, 10, 12]:
            max_day = 31
        elif self.month in [4, 6, 9, 11]:
            max_day = 30
        else:
            leap = (self.year % 400 == 0 or (self.year % 4 == 0 and self.year % 100 != 0))
            max_day = 29 if leap else 28
        self.day += 1

        if self.day > max_day:
            self.day = 1
            self.month += 1

            if self.month > 12:
                self.month = 1
                self.year += 1
s = input().strip()
d,m,y = map(int, s.split('/'))
date = Date(d,m,y)
if not date.is_valid():
    print('INVALID')
else:
    date.next_day()
    print(f"{date.day:02d}/{date.month:02d}/{date.year}")
```

```python
#w8a3
class Point:
    def __init__(self, x_value, y_value):
        self.x = x_value
        self.y = y_value
    
    def is_origin(self):
        return self.x == 0 and self.y == 0
    
    def on_which_axis(self):
        if self.x == 0:
            return "Nằm trên trục tung"
        if self.y == 0:
            return "Nằm trên trục hoành"
        else:
            return "Không nằm trên trục tọa độ"
    def dist(self):
        return ((self.x)**2 + (self.y)**2)**0.5
```

```python
#w8a5
class ShoppingCart:
  def __init__(self):
    self.items = []
  def add(self,name,price):
    self.items.append((name,price))
  def remove_item(self,name,price):
    try:
      self.items.remove((name, price))
      print("Đã xóa sản phẩm")
    except ValueError:
      print("Không tìm thấy sản phẩm")
  def check(self):
    if not self.items:
      return 'Giỏ hàng rỗng'
    else:
      return 'Giỏ hàng không rỗng'
  def sum_price(self):
    return sum(y for x,y in self.items)
  def show_all(self):
    if not self.items:
      print('Giỏ hàng rỗng')
    else:
      for name,price in self.items:
        print(f'{name} - {price}')
  def remove_all(self):
    self.items = []
    print('Đã xóa thành công')
  def menu(self):
    print('1:Thêm vào giỏ hàng')
    print('2:Xóa khỏi giỏ hàng')
    print('3:Kiểm tra giỏ hàng')
    print('4:Tổng giá trị')
    print('5:Hiển thị toàn bộ giỏ hàng')
    print('6:Xóa toàn bộ giỏ hàng')
    n=input('Chọn mục : ')
    while 1:
      if n == '1':
        self.add(input('Nhập tên sản phẩm '),int(input('Nhập giá sản phẩm ')))
      elif n == '2':
        self.remove_item(input('Nhập tên sản phẩm '),int(input('Nhập giá sản phẩm ')))
      elif n == '3':
        print(self.check())
      elif n == '4':
        print(self.sum_price())
      elif n == '5':
        self.show_all()
      elif n == '6':
        self.remove_all()
      elif n == 'yes':
        break
      n=input('Chọn mục : ')
s=ShoppingCart()
s.menu()
```

```python
#w8a6
class Rectangle:
    def __init__(self, w, h):
        self.width = w
        self.height = h
    def area(self):
        return self.width * self.height
    def perimeter(self):
        return 2*(self.width + self.height)
    def scale(self, k):
        self.width = self.width * k
        self.height = self.height * k

w, h, k = map(float, input().split())
rec = Rectangle(w, h)
rec.scale(k)
print(rec.area(), rec.perimeter())
```

```python
#w8a7
from math import*
class Fraction:
  def __init__(self,tu_1,mau_1,oper,tu_2,mau_2):
    self.tu_1=tu_1
    self.tu_2=tu_2
    if mau_1 == 0 or mau_2 == 0:
      raise ValueError('Mẫu khác 0')
    self.mau_2=mau_2
    self.mau_1=mau_1
    self.oper = oper
  def add(self):
    if self.mau_1 == self.mau_2:
      return f'{(self.tu_1+self.tu_2)//gcd(self.tu_1+self.tu_2,self.mau_1)}/{self.mau_1}'
    else:
      bcnn = lcm(self.mau_1,self.mau_2)
      self.tu_1 = bcnn // self.mau_1
      self.tu_2 = bcnn // self.mau_2
      return f'{(self.tu_1 + self.tu_2)//gcd(self.tu_1 + self.tu_2,bcnn)}/{bcnn // gcd(self.tu_1 + self.tu_2,bcnn)}'
  def sub(self):
    if self.mau_1 == self.mau_2:
      return f'{(self.tu_1-self.tu_2)//gcd(self.tu_1-self.tu_2,self.mau_1)}/{self.mau_1}'
    else:
      bcnn = lcm(self.mau_1,self.mau_2)
      self.tu_1 = bcnn // self.mau_1
      self.tu_2 = bcnn // self.mau_2
      return f'{(self.tu_1 - self.tu_2)//gcd(self.tu_1 - self.tu_2,bcnn)}/{bcnn // gcd(self.tu_1 - self.tu_2,bcnn)}'
  def mul(self):
    return f'{(self.tu_1 * self.tu_2)//gcd(self.tu_1 * self.tu_2,self.mau_1 * self.mau_2)}/{(self.mau_1 * self.mau_2)//gcd(self.tu_1 * self.tu_2,self.mau_1 * self.mau_2)}'
  def div(self):
    return f'{(self.tu_1 * self.mau_2)//gcd(self.tu_1 * self.mau_2,self.tu_2 * self.mau_1)}/{(self.mau_1 * self.tu_2)//gcd(self.tu_1 * self.mau_2,self.tu_2 * self.mau_1)}'
  def menu(self):
    if self.oper == '+':
      print(self.add())
    elif self.oper == '-':
      print(self.sub())
    elif self.oper == '*':
      print(self.mul())
    elif self.oper == '/':
      print(self.div())

tu_1,mau_1,oper,tu_2,mau_2 = map(str,input().split())
s=Fraction(int(tu_1),int(mau_1),oper,int(tu_2),int(mau_2))
s.menu()
```

```python
#w8a8
class BankAccount:
  def __init__(self,owner,balance):
    self.owner=owner
    self.balance=balance
  def deposit(self,dep):
    self.balance += dep
  def withdraw(self,wdr):
    if self.balance >= wdr:
      self.balance -= wdr
  def __str__(self):
    return ('%.2f' %self.balance)
owner,balance = map(str,input().split())
s=BankAccount(owner,float(balance))
n=int(input())
for _ in range(n):
  method, money = map(str,input().split())
  money = float(money)
  if method == 'WITHDRAW':
    s.withdraw(money)
  elif method == 'DEPOSIT':
    s.deposit(money)
print(s)
```

```python
#w8a9
class Student:
  def __init__(self,name):
    self.name = name
    self.scores = []
  def add(self,sub,score):
    self.scores.append((sub,score))
  def avg(self):
    b=[x[1] for x in self.scores]
    return round(sum(b)/len(b),2)
  def rank(self):
    if self.avg() >=8:
      return 'Excellent'
    elif 6.5<=self.avg() < 8:
      return 'Good'
    elif 5 <= self.avg() < 6.5:
      return 'Average'
    else:
      return 'Poor'
name=input()
s=Student(name)
n=int(input())
for _ in range(n):
  sub,score = map(str,input().split())
  s.add(sub,float(score))
print(f'{name} {s.avg():.2f} {s.rank()}')
```

```python
#w8a11
from math import*
class space:
  def __init__(self,a1,a2,a3,b1,b2,b3,c1,c2,c3,d1,d2,d3):
    self.a1=a1
    self.a2=a2
    self.a3=a3
    self.b1=b1
    self.b2=b2
    self.b3=b3
    self.c1=c1
    self.c2=c2
    self.c3=c3
    self.d1=d1
    self.d2=d2
    self.d3=d3
  def mp1(self):
    n1,n2,n3=self.a1-self.b1,self.a2-self.b2,self.a3-self.b3
    k1,k2,k3=self.a1-self.c1,self.a2-self.c2,self.a3-self.c3
    return [n2*k3-k2*n3,n3*k1-k3*n1,n1*k2-k1*n2]
  def mp2(self):
    n1,n2,n3=self.c1-self.b1,self.c2-self.b2,self.c3-self.b3
    k1,k2,k3=self.d1-self.c1,self.d2-self.c2,self.d3-self.c3
    return [n2*k3-k2*n3,n3*k1-k3*n1,n1*k2-k1*n2]
  def goc(self):
    a,b,c=self.mp1()
    x,y,z=self.mp2()
    tu = abs(a*x + b*y + c*z)
    mau= sqrt(a**2 + b**2 +c**2) * sqrt(x**2 +y**2 +z**2)
    return round(degrees(acos(tu/mau)),2)
a1,a2,a3 = map(float,input().split())
b1,b2,b3 = map(float,input().split())
c1,c2,c3 = map(float,input().split())
d1,d2,d3 = map(float,input().split())
s = space(a1,a2,a3,b1,b2,b3,c1,c2,c3,d1,d2,d3)
```

```python
#w8a12
class ComplexNumber:
  def __init__(self,real1,img1,real2,img2):
    self.real1 = real1
    self.img1 = img1
    self.real2 = real2
    self.img2 = img2
  def mul(self):
    return f'{self.real1*self.real2 - self.img1*self.img2} {self.real1*self.img2 + self.img1*self.real2}'
real1,img1 = map(int,input().split())
real2,img2 = map(int,input().split())
s = ComplexNumber(real1,img1,real2,img2)
print(s.mul())
```

```python
#w8a13
from math import*
class Triangle:
  def __init__(self,edge1,edge2,edge3):
    self.edge1=edge1
    self.edge2=edge2
    self.edge3=edge3
  def check(self):
    if (self.edge1 <= 0 or self.edge2 <= 0 or self.edge3 <= 0) or (self.edge1 + self.edge2 <= self.edge3 or self.edge1 + self.edge3 <= self.edge2 or self.edge3 + self.edge2 <= self.edge1):
      return False
    else: return True
  def dientich(self):
    if self.check():
      p = (self.edge1 + self.edge2 + self.edge3)/2
      s= sqrt(p*(p-self.edge1)*(p-self.edge2)*(p-self.edge3))
      return f'{s:.2f}'
    else:
      return "invalid"
a,b,c=map(float,input().split())
s = Triangle(a,b,c)
print(s.dientich())
```
